```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T = 30
Tneg= 7
t = np.arange(T) - Tneg
x = np.ones(T)
x[:Tneg] = 0
N = 8
h = np.zeros(T)
h[Tneg:Tneg+N] = 1. / N

y = np.convolve(h, x, "full")

fig, axarr = plt.subplots(1,3,figsize=(12,4), sharex=True, sharey=True)
axarr[0].stem(t, x)
axarr[0].set_title(r"input $x$")
axarr[0].set_xlabel(r"$n\to$")
axarr[1].stem(t, h)
axarr[1].set_title(r"filter $h$")
axarr[1].set_xlabel(r"$n\to$")
axarr[2].stem(t, y[Tneg:Tneg+len(t)], "C0")
axarr[2].set_title(r"output $y=h \ast x$")
axarr[2].set_xlabel(r"$n\to$")

plt.savefig("./images/cm01_transient_steady_state.svg")
plt.show()
```
